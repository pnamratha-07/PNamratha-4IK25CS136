```python
print("\nMissing values Before")
print(df.isnull().sum())
```

```python
df_mean = df.copy()
```

```python
mean_building_area = df_mean["BuildingArea"].mean()
```

```python
df_mean["BuildingArea"]=df_mean["BuildingArea"].fillna(mean_building_area)
print("\nMean Building Area:",mean_building_area)
print("Missing Building values after Mean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())
      
```

```python
df_median = df.copy()
median_year_built = df_median["YearBuilt"].median()
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(median_year_built)
print("\nMedian Year Built:",median_year_built)
```

```python
print("Missing Yearbuilt values after Median Imputation:")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill = df.copy()
df_ffill["Car"] = df_ffill["Car"].ffill()
print("\nMissing Car values after Forward Fill:")
print(df_ffill["Car"].isnull().sum())
```

```python
comparision=pd.DataFrame({"Original BuildingArea":df["BuildingArea"].head(15),"Mean Imputed BuildingArea":df_mean["BuildingArea"].head(15),"Original YearBuilt":df["YearBuilt"].head(15),"Median Imputed YearBuilt":df_median["YearBuilt"].head(15),"Original Car":df["Car"].head(15),"Foreward Filled Car":df_ffill["Car"].head(15)})
print("\nComparision of Imputation Methods:")
print(comparision)
```

```python
df_processed = df.copy()
council_mode = df_processed["CouncilArea"].mode()[0]
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(
    council_mode
)
print("CouncilArea Mode:",council_mode)
print("misssing categorical value after imputation")
print(df_processed["CouncilArea"].isnull().sum())
```

```python
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type","Method"],
    dtype=int
)
print("\nDataset after one-Hot Encoding:")
print(df_encoded.head())
```

```python
df_label = df.copy()
label_encoder = LabelEncoder()
df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)
print("\nLabel Encoding of CouncilArea:")
print(
    df_label[
        ["CouncilArea", "CouncilArea_Encoded"]
        ].head(10)
)    
```
